```python
def prim(g):
    n = len(g)
    vis = [0] * n
    vis[0] = 1

    for _ in range(n - 1):
        mn = float('inf')

        for i in range(n):
            for j in range(n):
                if vis[i] and not vis[j] and g[i][j] < mn:
                    mn = g[i][j]
                    u, v = i, j

        print(u, "-", v, ":", mn)
        vis[v] = 1

g = [
    [0, 2, 0, 6],
    [2, 0, 3, 8],
    [0, 3, 0, 1],
    [6, 8, 1, 0]
]

prim(g)
```
